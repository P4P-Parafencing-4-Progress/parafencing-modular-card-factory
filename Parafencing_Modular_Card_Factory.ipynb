# 4. Typographic Engine & Pure Logic Pipeline

Stage 4 enforces deterministic typography rules: eliminating orphan words via non-breaking spaces and auditing character budgets against 5x7 inch physical boundaries.


## 4.1 Typographic Orphan-Killer & Normalization Engine

### Dual-Interface Card: Typographic Orphan-Killer Engine
* **Role & Operational Purpose**: Eliminates typographic widows and orphan words by substituting trailing whitespace with non-breaking spaces (`&nbsp;`). Enforces strict zero-em-dash compliance.
* **Inputs & Dependencies**: Raw string text from drill parameters.
* **Outputs Produced**: Cleaned, orphan-free HTML-safe text strings.
* **Safe Patch Guidance**: Pure Python regex function. Never bypass this filter before compiling HTML.


````python
# [Code 4.1] Deterministic Typographic Orphan-Killer & Normalizer
import re

def kill_orphans(text: str) -> str:
    """
    Replaces the space before the last word of a sentence or bullet item
    with a non-breaking space (&nbsp;) to prevent single-word orphan wrap.
    Strictly removes any em dashes or double-hyphen substitutes.
    """
    if not text or not isinstance(text, str):
        return ""
    
    # Strip any accidental em dashes or en dashes
    clean_text = text.replace("—", ": ").replace("–", "-").replace("--", "-")
    
    # Clean multiple spaces
    clean_text = re.sub(r'\s+', ' ', clean_text).strip()
    
    # Replace the last space before the final word with &nbsp;
    parts = clean_text.rsplit(' ', 1)
    if len(parts) == 2:
        return f"{parts[0]}&nbsp;{parts[1]}"
    return clean_text

def normalize_drill_record(drill: dict) -> dict:
    """Applies typographic normalization across all textual fields."""
    normalized = drill.copy()
    normalized["title_clean"] = kill_orphans(drill["title"])
    normalized["subtitle_clean"] = kill_orphans(drill["subtitle"])
    normalized["tactical_phase_clean"] = kill_orphans(drill["tactical_phase"])
    normalized["prerequisites_clean"] = kill_orphans(drill["prerequisites"])
    normalized["equipment_clean"] = kill_orphans(drill["equipment"])
    
    normalized["steps_clean"] = [kill_orphans(s) for s in drill.get("steps", [])]
    normalized["coach_cues_clean"] = [kill_orphans(c) for c in drill.get("coach_cues", [])]
    normalized["self_audit_clean"] = [kill_orphans(a) for a in drill.get("self_audit", [])]
    return normalized

SAMPLE_RAW = "Execute Rule o.102 mesurage using bent elbow crease to opponent guard center."
SAMPLE_CLEAN = kill_orphans(SAMPLE_RAW)
print(f"Orphan-Killer Verification:")
print(f"  Raw:   '{SAMPLE_RAW}'")
print(f"  Clean: '{SAMPLE_CLEAN}'")
````

## 4.2 Layout Budget & Fencing Schema Validator

### Dual-Interface Card: Layout Budget & Fencing Schema Validation
* **Role & Operational Purpose**: Validates that all drill cards adhere to character budgets (ensuring zero 5x7 page overflow) and satisfy the 6 invariant sections of the French *Fiche Pédagogique*.
* **Inputs & Dependencies**: Normalized drill dictionary.
* **Outputs Produced**: Boolean validation pass with detailed character counts.
* **Safe Patch Guidance**: If a drill fails character budgeting, trim verbose phrasing. Never increase 5x7 container dimensions.


````python
# [Code 4.2] Pure Python Layout Budget & Schema Gate
BUDGET_LIMITS = {
    "title_max": 65,
    "subtitle_max": 120,
    "step_max": 140,
    "cue_max": 140,
    "audit_max": 140
}

def validate_fencing_schema(drill: dict) -> tuple:
    issues = []
    
    # 1. Invariant section presence check (French Master Fiche Pedagogique)
    required_keys = ["module_code", "category", "title", "steps", "coach_cues", "self_audit"]
    for k in required_keys:
        if not drill.get(k):
            issues.append(f"Missing mandatory section: {k}")
            
    # 2. Length budgeting checks (Prevents 5x7 footer cutoff)
    if len(drill.get("title", "")) > BUDGET_LIMITS["title_max"]:
        issues.append(f"Title exceeds budget ({len(drill['title'])}/{BUDGET_LIMITS['title_max']})")
        
    if len(drill.get("subtitle", "")) > BUDGET_LIMITS["subtitle_max"]:
        issues.append(f"Subtitle exceeds budget ({len(drill['subtitle'])}/{BUDGET_LIMITS['subtitle_max']})")
        
    for i, s in enumerate(drill.get("steps", [])):
        if len(s) > BUDGET_LIMITS["step_max"]:
            issues.append(f"Step {i+1} exceeds budget ({len(s)}/{BUDGET_LIMITS['step_max']})")
            
    for i, c in enumerate(drill.get("coach_cues", [])):
        if len(c) > BUDGET_LIMITS["cue_max"]:
            issues.append(f"Coach cue {i+1} exceeds budget ({len(c)}/{BUDGET_LIMITS['cue_max']})")
            
    for i, a in enumerate(drill.get("self_audit", [])):
        if len(a) > BUDGET_LIMITS["audit_max"]:
            issues.append(f"Self-audit {i+1} exceeds budget ({len(a)}/{BUDGET_LIMITS['audit_max']})")
            
    is_valid = len(issues) == 0
    return is_valid, issues

test_valid, test_issues = validate_fencing_schema(ACTIVE_DRILL)
if test_valid:
    print(f"Validation Gate: 100% PASS - Drill adheres strictly to 5x7 layout budget.")
else:
    print(f"Validation Gate: WARNING - {len(test_issues)} budget issue(s) detected: {test_issues}")
````

## 5.2 Deterministic PDF & PNG Exporter

### Dual-Interface Card: Automated Headless PDF & High-Res PNG Exporter
* **Role & Operational Purpose**: Executes headless Playwright/Chromium rendering to compile exact 5x7 inch print-ready PDFs (`output/pdf/`) and high-resolution PNG card images (`output/png/`).
* **Inputs & Dependencies**: Compiled HTML decks, Playwright or WeasyPrint.
* **Outputs Produced**: Print-ready PDF files and standalone PNG images for each card face.
* **Safe Patch Guidance**: If headless browser is absent, the user can click the toolbar buttons inside the HTML files to export with 1 click.


````python
# [Code 5.2] Headless PDF & High-Res PNG Exporter
pdf_light = OUTPUT_PDF / "parafencing_cards_usopc_light.pdf"
pdf_dark = OUTPUT_PDF / "parafencing_cards_mobile_dark.pdf"

exported_files = []

if HAS_PLAYWRIGHT:
    try:
        from playwright.sync_api import sync_playwright
        print("Executing headless Playwright rendering for PDF and PNG export...")
        with sync_playwright() as p:
            browser = p.chromium.launch()
            for theme_name, html_path, pdf_path in [
                ("light", path_light_html, pdf_light),
                ("dark", path_dark_html, pdf_dark)
            ]:
                page = browser.new_page(viewport={"width": 1500, "height": 2100}, device_scale_factor=2)
                page.goto(f"file://{html_path.resolve()}")
                
                # Render 5x7 PDF
                page.pdf(path=str(pdf_path), width="5in", height="7in", print_background=True, margin={"top":"0","bottom":"0","left":"0","right":"0"})
                exported_files.append(str(pdf_path))
                
                # Render PNG for each card page
                card_elements = page.query_selector_all(".card-page")
                for c_idx, card_el in enumerate(card_elements):
                    face_tag = "side_a" if c_idx % 2 == 0 else "side_b"
                    card_num = (c_idx // 2) + 1
                    png_path = OUTPUT_PNG / f"parafencing_{theme_name}_card_{card_num}_{face_tag}.png"
                    card_el.screenshot(path=str(png_path))
                    exported_files.append(str(png_path))
                    
            browser.close()
            print(f"Playwright successfully generated {len(exported_files)} PDF and PNG deliverables.")
    except Exception as e:
        print(f"Playwright rendering notice: {e}")
elif HAS_WEASYPRINT:
    try:
        import weasyprint
        print("Executing WeasyPrint PDF compilation...")
        weasyprint.HTML(filename=str(path_light_html)).write_pdf(str(pdf_light))
        weasyprint.HTML(filename=str(path_dark_html)).write_pdf(str(pdf_dark))
        exported_files.extend([str(pdf_light), str(pdf_dark)])
        print(f"WeasyPrint generated {len(exported_files)} PDF files.")
    except Exception as e:
        print(f"WeasyPrint rendering notice: {e}")
else:
    print("Notice: Automated headless renderer not detected. Use the top action bar in the HTML file to save as PDF or PNG with 1 click.")

print("Stage 5 Complete: All deliverables assembled.")
````
